```python
import random

# Set random seed for reproducibility
random.seed(42)

# Define possible product attributes

brands = ["Aster", "Northline", "Vista", "ClearView", "Urban Optics", "Luma"]

shapes = ["Round", "Square", "Rectangle", "Oval", "Cat Eye", "Aviator"]

materials = ["Acetate", "Metal", "Titanium", "Mixed"]

colours = ["Black", "Brown", "Tortoise", "Gold", "Silver", "Blue", "Transparent"]

genders = ["Women", "Men", "Unisex"]

availability_options = ["In Stock", "Out of Stock"]

# Create an empty list
products = []

# Generate 150 products
for i in range(1, 151):

    shape = random.choice(shapes)

    product = {
        "product_id": f"GL{i:04d}",
        "product_name": f"{random.choice(['Classic', 'Modern', 'Essential', 'Studio', 'Everyday', 'Premium'])} {shape} Frame {i}",
        "brand": random.choice(brands),
        "price": round(random.uniform(49.99, 249.99), 2),
        "frame_colour": random.choice(colours),
        "frame_shape": shape,
        "frame_material": random.choice(materials),
        "gender": random.choice(genders),
        "availability": random.choice(availability_options)
    }

    products.append(product)

print("150 synthetic glasses products created!")
```

```python
# Create six HTML pages

for page_number in range(1, 7):

    # Select 25 products for each page
    start_index = (page_number - 1) * 25
    end_index = start_index + 25

    page_products = products[start_index:end_index]

    # Create HTML product cards
    product_cards = []

    for product in page_products:

        card = f"""
        <div class="product-card" data-product-id="{product['product_id']}">

            <h2 class="product-name">{product['product_name']}</h2>

            <span class="brand">{product['brand']}</span>

            <span class="price">£{product['price']:.2f}</span>

            <span class="frame-colour">{product['frame_colour']}</span>

            <span class="frame-shape">{product['frame_shape']}</span>

            <span class="frame-material">{product['frame_material']}</span>

            <span class="gender">{product['gender']}</span>

            <span class="availability">{product['availability']}</span>

        </div>
        """

        product_cards.append(card)

    # Combine product cards
    cards_html = "".join(product_cards)

    # Create the complete HTML page
    html_content = f"""
    <!DOCTYPE html>
    <html>
    <head>
        <meta charset="UTF-8">
        <title>Frame & Lens Catalogue</title>
    </head>

    <body>

        <h1>Frame & Lens Practice Catalogue</h1>

        <div class="catalogue">

            {cards_html}

        </div>

        <p>Page {page_number}</p>

    </body>
    </html>
    """

    # Save the HTML page
    file_path = f"html_pages/page_{page_number}.html"

    with open(file_path, "w", encoding="utf-8") as file:
        file.write(html_content)

    print(f"Page {page_number} created successfully!")
```

```python
from http.server import ThreadingHTTPServer, SimpleHTTPRequestHandler
from functools import partial
from threading import Thread

# Set the folder containing our HTML files
html_folder = os.path.abspath("html_pages")

# Create the server handler
handler = partial(SimpleHTTPRequestHandler, directory=html_folder)

# Create the local server
server = ThreadingHTTPServer(("127.0.0.1", 8000), handler)

# Start the server in the background
server_thread = Thread(target=server.serve_forever, daemon=True)

server_thread.start()

print("Local web server started!")
print("Practice catalogue: http://127.0.0.1:8000/page_1.html")
```

```python
from selenium.webdriver.chrome.options import Options

# Configure Chrome

chrome_options = Options()

# Run Chrome in the background
chrome_options.add_argument("--headless=new")

# Disable GPU
chrome_options.add_argument("--disable-gpu")

# Create the browser
driver = webdriver.Chrome(options=chrome_options)

print("Selenium browser successfully configured!")
```

```python
# Open the first catalogue page

driver.get("http://127.0.0.1:8000/page_1.html")

# Print the page title
print("Page title:", driver.title)

# Print the first 500 characters of the page HTML
print(driver.page_source[:500])
```

```python
# Get the HTML from Selenium
html = driver.page_source

# Parse the HTML using BeautifulSoup
soup = BeautifulSoup(html, "html.parser")

# Find all product cards
product_cards = soup.select(".product-card")

print("Products found:", len(product_cards))
```

```python
# Create an empty list
page_data = []

# Extract data from every product card
for card in product_cards:

    product = {
        "product_id": card.get("data-product-id"),
        "product_name": card.select_one(".product-name").get_text(strip=True),
        "brand": card.select_one(".brand").get_text(strip=True),
        "price": card.select_one(".price").get_text(strip=True).replace("£", ""),
        "frame_colour": card.select_one(".frame-colour").get_text(strip=True),
        "frame_shape": card.select_one(".frame-shape").get_text(strip=True),
        "frame_material": card.select_one(".frame-material").get_text(strip=True),
        "gender": card.select_one(".gender").get_text(strip=True),
        "availability": card.select_one(".availability").get_text(strip=True)
    }

    page_data.append(product)

# Convert extracted data into a DataFrame
page_df = pd.DataFrame(page_data)

page_df.head()
```

```python
# Store all collected products
all_products = []

# Visit six pages
for page_number in range(1, 7):

    # Open the page with Selenium
    page_url = f"http://127.0.0.1:8000/page_{page_number}.html"

    driver.get(page_url)

    # Allow the page to load
    time.sleep(0.3)

    # Retrieve page HTML
    html = driver.page_source

    # Parse the HTML
    soup = BeautifulSoup(html, "html.parser")

    # Find product cards
    cards = soup.select(".product-card")

    # Extract data from each product
    for card in cards:

        product = {
            "product_id": card.get("data-product-id"),
            "product_name": card.select_one(".product-name").get_text(strip=True),
            "brand": card.select_one(".brand").get_text(strip=True),
            "price": card.select_one(".price").get_text(strip=True).replace("£", ""),
            "frame_colour": card.select_one(".frame-colour").get_text(strip=True),
            "frame_shape": card.select_one(".frame-shape").get_text(strip=True),
            "frame_material": card.select_one(".frame-material").get_text(strip=True),
            "gender": card.select_one(".gender").get_text(strip=True),
            "availability": card.select_one(".availability").get_text(strip=True),
            "page_number": page_number,
            "source_website": "Local Practice Catalogue",
            "data_type": "Synthetic practice data"
        }

        all_products.append(product)

    print(f"Page {page_number}: {len(cards)} products collected")

print("\nTotal products collected:", len(all_products))
```
